````python
# 1. write a python program to print 'Hello, World!'
print('Hello, World!')
````

````python
# 2. create a program that print users name and age
user_name = input("Enter your name: ")
user_age = input("Enter your age: ")

print(f"Hello {user_name}, you are {user_age} years old!")
````

````python
# 3. create a program to checked whether user's input is either even or odd
user_input = int(input('Enter a number '))

if user_input % 2 == 0:
    print('Even')
else:
    print('Odd')
````

````python
# 4. creating a program that prints multiplication table of a number (from 1 -12)
# get a number from user
num = int(input('Enter a number: '))
print(f"Multiplication Table for {num}")

# then loop from 1 to 12
for i in range(1,13):
    print(f"{num} x {i} = {num * i}")
````

````python
# 5. a program that calculates the factorial of a number using loop
num = int(input("Enter a number to calculate its factorial: "))
factorial = 1

if  num < 0:
    print("Factorial is not defined for negative numbers.")
elif num == 0:
    print("The factorial of 0 is 1.")
else:
    for i in range(1, num + 1):
        factorial *= i
    print(f"The factorial of {num} is {factorial}.")
````

````python
# 6. Write a program that finds the largest number in a given list.
numbers = [5, 15, 10, 25, 3, 7, 8, 2]
largest = numbers[0]
for num in numbers:
    if num > largest:
        largest = num
print(f"The largest number is {largest}")
````

````python
# 7. #Create a Python program that counts how many vowels are in a given string.
text = 'Yussuf is going to be a great data scientist'
vowels = 'aeiouAEIOU'
count = 0

for char in text:
    if char in vowels:
        count += 1

print(f"The number of vowels in the given string is: {count}")
````

````python
# 8. write a function that takes two numbers and returns their sum, difference, and product
def calculate_numbers(num1, num2):
    sum = num1 + num2
    difference = num1 - num2
    product = num1 * num2   
    return sum, difference, product

num1 = int(input("Enter the first number: "))
num2 = int(input("Enter the second number: "))

sum, difference, product = calculate_numbers(num1, num2)

print("Sum:", sum)
print("Difference:", difference)
print("Product:", product)
````

````python
# Create a program that reverses a string without using slicing ([::-1])
text = input('Enter a string: ')
reversed_text = "".join(reversed(text))
print(reversed_text)
````

````python
# Write a program that generates the first 10 numbers of the Fibonacci sequence.
a, b = 0, 1 

print("First 10 numbers of the Fibonacci sequence:")
for _ in range(10):
    print(a, end=" ")
    a, b = b, a + b
````

````python
#Student Grading System  - #Write a program to accept marks for 5 subjects, 
# calculate the average, and display the grade based on:
#A: 80–100, B: 60–79, C: 40–59, F: < 40

mark1 = int(input("Enter marks for subject 1: "))
mark2 = int(input("Enter marks for subject 2: "))
mark3 = int(input("Enter marks for subject 3: "))
mark4 = int(input("Enter marks for subject 4: "))
mark5 = int(input("Enter marks for subject 5: "))   

avg = (mark1 + mark2 + mark3 + mark4 + mark5) / 5

if avg >= 80:
    grade = 'A'
elif avg >= 60:
    grade = 'B'
elif avg >= 40:
    grade = 'C'             
else:
    grade = 'F'

print("Average is:", avg)
print("Grade is:", grade) 
````

````python
# simple calculator
# create a program that can perform addition, substraction, multiplication 
# and division based on users input

x = float(int(input('Enter a num: ')))
y = float(int(input('Enter another num: ')))

# choose an operation
operation = input('Enter an operation (+, -, *, /): ')

if operation == '+':
    print("The result is:", x + y)
elif operation == '-':
    print("The result is:", x - y)
elif operation == '*':
    print("The result is:", x * y)
elif operation == '/':
    if y != 0:
        print("The result is:", x / y)
    else:
        print("Error: Division by zero")
else:
    print("Error: Invalid operation")
````

````python
# shopping cart
# Create a program where a user can add items to a shopping cart (dictionary with item names and prices), then display the total cost
cart = {}

while True:
    item = input("Enter item name (or 'done' to finish): ")
    if item.lower()== 'done':
        break
    else:
        price = float(input(f"Enter price for {item}: "))
        cart[item] =  price

#total_cost = sum(cart.values())

total = 0

for item, price in cart.items():
    total = total + price

print("Cart:", cart)
print("Total is", total)
````

````python
#Guessing Game
#Write a program where the computer randomly selects a number between 1–20, and the user must guess it. Give hints if the guess is too high or too low.

import random

num = random.randint(1, 20)
while True:
    guess = int(input("Guess a number between 1 and 20: "))

    print("You guessed:", guess)   # show user’s input
    if guess < num:
        print("Too low! Try again.")

    elif guess > num:
        print("Too high! Try again.")

    else:
        print("Correct! The number was: ", num)
        break
````

````python
#File Handling
#Write a program that accepts a list of student names from the user, saves them in a text file, and then reads and displays them.
students = []
done = False

while not done:
    name = input("Enter student name (or 'done' to finish) ")
    if name == 'done':
        done = True
        break
    else:
        students.append(name)

#save the names in a text file
f = open("students.txt", "w")
for student in students:
    f.write(student + "\n")
f.close()

# reading the file
f = open("students.txt", "r")
for line in f:
    print(line.strip())
f.close()
````

**6. What is the difference between local and global variables? Provide an example.**
local: a variable declared inside a function and is only accessible there
global: a variable declared outside a function and can be accessed anywhere in the program

```python
total = 0  # global

def add_to_total(amount):
    global total
    total += amount

def calculate_tax(price):
    tax = price * 0.1  # local — only exists inside this function
    return tax

add_to_total(100)
add_to_total(50)
print(total)  # Output: 150

print(tax)    # Error! 'tax' doesn't exist outside calculate_tax()
```
