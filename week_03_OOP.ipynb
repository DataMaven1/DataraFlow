````python
# assignment 1: JSON File Handling - Write a program that stores a list of student records (name, age, grade) into a JSON file,
#  and then reads the file back into a Python object.

import json
# creating list of students records
students = [
    {"name":"Yussuf", "age":25, "grade":"A"},
    {"name":"Bolaji", "age":20, "grade":"B"},
    {"name":"Winner", "age":30, "grade":"A"},
    {"name":"Divine", "age":19, "grade":"C"}
]

# storing the records into a json file - write mode
with open("students.json", "w") as file:
    json.dump(students, file)

# reading the data from the file into a py object - read mode
with open("students.json", "r") as file:
    records = json.load(file)

# output the data to confirm it
print(records)
````

**2: Math Quiz - Build a program that asks the user 5 random math question (addition, subtraction, multiplication, division) and keeps score of correct answers.**

````python
# assignemnt 2: Math Quiz - Build a program that asks the user 5 random math questions
# (addition, subtraction, multiplication, division) and keeps score of correct answers.

import random

score = 0 #keeping track of the correct response

# start a loop that runs 5 times
for i in range(5):
    # this will Pick two random numbers between 1 and 10
    num1 = random.randint(1,10)
    num2 = random.randint(1,10)
    operation = random.choice(["+", "-", "*", "/"])

    #calculating the correct answers
    if operation == "+":
        correct_answer = num1 + num2
    elif operation == "-":
        correct_answer = num1 - num2
    elif operation == "*":
        correct_answer = num1 * num2
    else: #division
        correct_answer = round(num1 / num2, 2)

    # catches error -if the user types a letter instead of a number
    try:
        user_ans = float(input(f"What is {num1} {operation} {num2} "))
    except ValueError:
        print("This is not a valid number")
        user_ans = None

    # check if the user is correct
    if user_ans == correct_answer:
        print("Correct")
        score += 1 # add +1 to the score
    else:
        print(f"Incorrect answer. Correct answer is {correct_answer}")

print(f"You got {score} out of 5 correct!")
````

**3.Error Handling System -Create a program that asks the user for a number and divides 100 by that number. Use `try/except` to handle invalid input and division by zero**

````python
# assignment 3.Error Handling System -Create a program that asks the user for a number and divides 100 by that number.
#  Use `try/except` to handle invalid input and division by zero.

try:
    # ask for user input
    num = float(input("Enter a num: "))
    #then divide 100 by input
    result = 100 / num
    print("Result is: ", result)

#handles invalid number - like letters
except ValueError:
    print("Invalid input. Enter a num")
# handles division by zero
except ZeroDivisionError:
    print("Error, Cannot divide by zero")

````

**4.Date Calculator - Write a program that takes a user’s birthday as input and tells them how many days are left until their next birthday.**

````python
# assignment 4.Date Calculator - Write a program that takes a user’s birthday as input and
# tells them how many days are left until their next birthday.
from datetime import date


today = date.today()

try:
    # user input for month and day
    birth_month = int(input("Enter your birth month(1-12): "))
    birth_day = int(input("Enter your birth day(1-31): "))

    #a date obj to get there next birthday
    next_birthday = date(today.year, birth_month, birth_day)

    # this check if there birthday has already passed then move to next year
    if today > next_birthday:
        next_birthday = date(today.year + 1, birth_month, birth_day)

    #calculate the diffrence in days
    days_left = (next_birthday - today).days

    # if birthday is today, greet them or print remaining days
    if days_left == 0:
        print("Happy Bithday")
    else:
        print(f"Your next birthday is in {days_left} days")

except ValueError:
    print("Invalid date entered or number")

````

**3. Explain the purpose of the `math.pi` constant and give an example of its use**.  

the `math.pi` is a constant in math module, it's a mathematical representation of π (3.14159...). which means we don't need to manually type the value, python gives the value through math.pi - It is used to compute the area, circumference, and volume of circles, spheres, and cylinders(used in geometry, trigonometry)

example - volume of a cylinder

````python
# volume of a cylinder
import math

# Dimensions of the cylinder
radius = 3.0
height = 10.0

# Calculate volume
volume = math.pi * (radius ** 2) * height

print(f"The volume of the cylinder is: {volume:.2f} cubic units")
````
