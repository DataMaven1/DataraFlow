````python
import pandas as pd
import sys, pathlib
_p = pathlib.Path.cwd()
while _p != _p.parent and not (_p / "db" / "dataraflow_db.py").exists():
    _p = _p.parent
sys.path.insert(0, str(_p))
from db.dataraflow_db import run_query
````

````python
london = run_query("SELECT * FROM week06.london_2014")
london.head()
````

````python
# City label -> table name in the week06 schema
tables = {
    "Beijing": "beijing_pek_2014",
    "Brasilia": "brasilia_bsb_2014",
    "Cape Town": "capetown_cpt_2014",
    "Delhi": "delhi_del_2014",
    "London": "london_2014",
    "Moscow": "moscow_svo_2014",
}

datasets = {}  # city name -> DataFrame
for city_name, table in tables.items():
    try:
        datasets[city_name] = run_query(f"SELECT * FROM week06.{table}")
        datasets[city_name].to_csv(f"{table}.csv", index=False)   # offline backup
        print(f"Loaded {city_name}: {datasets[city_name].shape}")
    except Exception as e:
        print(f"Error loading {city_name}: {str(e)[:80]}")
````

````python
for city, df in datasets.items():
    print(f"\n=== {city} ===")
    # rows x columns
    print("Shape:", df.shape)

    # how many columns of each type
    print(df.dtypes.value_counts())
    # missing values per column
    missing = df.isna().sum()
    print("Columns with missing values:")
    print(missing[missing > 0])
````

````python
for city, df in datasets.items():
    # Remove leading/trailing spaces
    df.columns = df.columns.str.strip().str.replace("<br />", "", regex=False)

    # London to  date columnwhich was GMT before,
    df.rename(columns={"GMT": "date", "Date": "date"}, inplace=True)

    # Dates are text like '2014-1-1'; convert them to real dates
    df["date"] = pd.to_datetime(df["date"])

    # Add a city label so the six tables can be stacked later
    df["city"] = city

    # Missing event = no event recorded
    #df["Events"] = df["Events"].fillna("No event")

# Check one city's cleaned column names
print(datasets["London"].columns.tolist())
````

````python
# Summary statistics for key columns, one city at a time
cols = ["Mean TemperatureC", "Mean Humidity", "Precipitationmm"]
for city, df in datasets.items():
    print(f"\n=== {city} ===")
    print(df[cols].describe().round(1))
````

````python
# Stack the six cleaned DataFrames into one long table
all_cities = pd.concat(datasets.values(), ignore_index=True)

# Add a month number (1-12) for monthly questions later
all_cities["month"] = all_cities["date"].dt.month

# Checks: expect (2190, 25) = 2190 rows, and the 24 columns plus "month"
print(all_cities.shape)

# Each city should appear exactly 365 times
print(all_cities["city"].value_counts())

# Dates should now be real datetimes, not text
print(all_cities["date"].dtype)
````

````python
# How many days had each event, side by side for the six cities
print(pd.crosstab(all_cities["Events"].fillna("No events"), all_cities["city"]))

print("-" * 50)
# Days with rain anywhere in the event (includes Fog-Rain, etc.)
rain_days = all_cities[all_cities["Events"].str.contains("Rain")].groupby("city").size()
print(rain_days.sort_values(ascending=False))
````

````python
# Mean temperature and humidity for each event type in each city
by_event = (all_cities
            .groupby(["city", "Events"], dropna=False)[["Mean TemperatureC", "Mean Humidity"]]
            .mean()
            .round(1))
print(by_event)
````

````python
# 1. Average temperature per city, warmest first
print(all_cities.groupby("city")["Mean TemperatureC"].mean().round(1).sort_values(ascending=False))

# 2. Total rainfall per city
print(all_cities.groupby("city")["Precipitationmm"].sum().round(1).sort_values(ascending=False))

# 3. Average temperature by month for each city (a seasonal pattern)
monthly = all_cities.pivot_table(index="month", columns="city",
                                 values="Mean TemperatureC", aggfunc="mean").round(1)
print(monthly)

# 4. Hottest day for each city
idx = all_cities.groupby("city")["Max TemperatureC"].idxmax()

print(all_cities.loc[idx, ["city", "date", "Max TemperatureC"]])
````

````python
# Moscow's precipitation: how many days are exactly zero?
print((datasets["Moscow"]["Precipitationmm"] == 0).sum(), "zero days")

# Total rainfall per city, leaving out Moscow (its column is all zeros)
total_mm = all_cities.groupby("city")["Precipitationmm"].sum().drop("Moscow")
print(total_mm.round(1).sort_values(ascending=False))

# Delhi: does it look right?
print(datasets["Delhi"]["Mean TemperatureC"].describe())
print(monthly[["Delhi", "Moscow"]])
````

## The report, findings

**1. General summary**
The data has daily weather for six cities: Beijing, Brasilia, Cape Town, Delhi, London and Moscow. Each city has 365 rows (one year of days) and 23 columns, covering temperature, humidity, visibility, wind, rainfall and weather events. I combined them into one table of 2,190 rows.

**2. Data structure and quality**
- **Structure:** every city has the same 365 × 23 shape. Most columns are numbers. For Beijing, Brasilia, Delhi and London that is 17 whole-number, 3 decimal and 3 text columns. Cape Town and Moscow have 14 whole-number, 6 decimal and 3 text columns, because their visibility columns have gaps.
- **Missing data:** it is common in every city, mostly in the same three columns.
  - `Max Gust SpeedKm/h` is missing on most days: 363 of Delhi's 365, 329 of Brasilia's.
  - `CloudCover` is missing on 224 of Beijing's days and on only 10 of London's.
  - `Events` is blank on many days (243 in Beijing, 114 in London).
  - Cape Town (2 days) and Moscow (32 days) also have gaps in visibility.
- **Messy names:** some column names had leading spaces, one had a stray `<br />`, and London's date column is `GMT` while the others column is `Date`.
- **What I cleaned:** I fixed the column names, used one name for the date, converted dates to real dates, and I left blank Events as missing values and counted them as days with no event recorded.".
- **Two things that look wrong:**
  - Moscow's rainfall is exactly 0 on all 365 days, yet 114 days have rain in the events column. I treated the column as unrecorded and left Moscow out of rainfall comparisons.
  - Delhi's temperatures look odd for Delhi (a January average of -1.9 °C, some snow entries). I can't tell if the file is wrong, so its source should be checked.

**3. Weather events**
"No event" is the most common entry in five of the six cities. Here are the main events in each:
- **Beijing:** Rain (44 days), Rain-Thunderstorm (33), Fog (28).
- **Brasilia:** Rain-Thunderstorm (75), Rain (41), Thunderstorm (30).
- **Cape Town:** Rain (87), Fog (38).
- **Delhi:** Rain (51), Thunderstorm (13).
- **London:** Rain (202) is bigger than "No event" (114), so London is the exception.
- **Moscow:** Snow (73) and Rain (70).

**4. Averages by event**
- Compared with no-event days, fog and snow days are colder and more humid. In London, fog averages 10.5 °C at 77.3% humidity against 14.4 °C and 66.2% on no-event days. In Moscow, snow days average -6.2 °C at 83.3%.
- Rain-thunderstorm days are warmer (London 16.4 °C, Moscow 18.3 °C).
- The humidity pattern isn't the same everywhere. In Beijing, snow days (58.3%) are less humid than rain days (62.0%).

**5. Temperature and rainfall by city**
- Average temperature, warmest first: Brasilia 22.9, Cape Town 17.6, Delhi 13.7, Beijing 13.4, London 12.3, Moscow 6.0.
- Moscow goes from -9.6 °C in January to 19.5 °C in July. Cape Town is reversed, warmest in February and coolest in July, as expected for the Southern Hemisphere.
- Total rainfall (five cities, Moscow left out): Brasilia 751.6 mm, London 503.1, Cape Town 428.2, Beijing 405.9, Delhi 225.3.

**6. Conclusion**
The cities show clear seasons, and the Southern Hemisphere cities are reversed. Brasilia is the warmest and has the most rain, Moscow is the coldest, and London has the rainiest event pattern. The Delhi results and Moscow's rainfall should be treated with caution.

### SQL Task 2 · Filtering & NULLs
For **London** (`week06.london_2014`), return **one row** with three counts:
1. `no_event_days`: days with **no event recorded**
2. `rain_days`: days whose event mentions **Rain** anywhere (e.g. `Rain`, `Fog-Rain`, `Rain-Thunderstorm`)
3. `wet_but_no_event_days`: days with **no event recorded** but `Precipitationmm > 0`

⚠️ The events column name starts with a **space**: `[ Events]`. And remember, `= NULL` never matches.
*Revisit: §1.2, §3.2, §3.3.*

````python
SQL_02 = """
    SELECT
        SUM(CASE WHEN [ Events] is NULL THEN 1 ELSE 0 END) AS no_event_days,
        SUM(CASE WHEN [ Events] LIKE '%Rain%' THEN 1 ELSE 0 END) AS rain_days,
        SUM(CASE WHEN [ Events] IS NULL AND [Precipitationmm] > 0 THEN 1 ELSE 0 END) AS wet_but_no_event_days
    FROM week06.london_2014
"""
answer("s02", SQL_02)
````

### SQL Task 3 · CASE & conditional aggregation
From `week16.customer_churn_prediction`, for each `Contract_Type` return: number of `customers`, number
`churned` (`Churn = 1`), and `churn_rate_pct` (percentage, **rounded to 1 decimal place**).

**Return:** `Contract_Type`, `customers`, `churned`, `churn_rate_pct`, sorted by `churn_rate_pct` (highest first).
⚠️ Watch out for the integer-division trap. *Revisit: §4.4, §5.2.*

````python
SQL_03 = """
    SELECT
        Contract_Type,
        Count(*) AS customers,
        SUM(CASE WHEN Churn = 1 THEN 1 ELSE 0 END) AS churned,
        ROUND(100.0 * AVG(CASE WHEN Churn = 1 THEN 1.0 ELSE 0 END), 1) AS churn_rate_pct
    FROM week16.customer_churn_prediction
    GROUP BY Contract_Type
    ORDER BY churn_rate_pct DESC
"""
answer("s03", SQL_03)
````

### SQL Task 4 · GROUP BY + HAVING
From `week09.comtrade_milk_uk_monthly_14`, find the partner countries the UK **exported** milk to in

**Return:** `Partner`, `total_exports_usd`, sorted by `total_exports_usd` (highest first).
⚠️ Exclude the `'World'` total row. *Revisit: §6.2.*

### SQL Task 5 · JOIN across two sources
Join WHO TB data (`week06.who_pop_tb_all`) with World Bank population (`week08.wb_pop_2013`) on the country
name. Compute **TB deaths per 100,000 people** using the **World Bank** population (`[SP.POP.TOTL]`),
**rounded to 1 decimal place**.

**Return:** the **top 5** countries: `Country`, `tb_deaths_per_100k`, highest first.
*Revisit: §7.2–7.3.*

````python
SQL_05 = """
    SELECT TOP 5
        w.Country,
        ROUND(100000.0 * w.[TB deaths] / NULLIF(p.[SP.POP.TOTL], 0), 1) AS tb_deaths_per_100k
    FROM week06.who_pop_tb_all as w
    JOIN week08.wb_pop_2013 as p ON p.Country = w.Country
    ORDER BY tb_deaths_per_100k DESC
"""
answer("s05", SQL_05)
````

### SQL Task 7 · UNION ALL
Stack the **six** city weather tables in `week06` and compute each city's **average `Mean TemperatureC`**
for 2014, **rounded to 1 decimal place**. Use exactly these city labels:
`'Beijing'`, `'Brasilia'`, `'Cape Town'`, `'Delhi'`, `'London'`, `'Moscow'`.

**Return:** `city`, `avg_mean_temp_c`, sorted by `city` (A→Z).
⚠️ `Mean TemperatureC` is stored as a whole number, so watch the integer-division trap. *Revisit: §8.1–8.2.*

````python
SQL_07 = """
    SELECT
        'London' AS city, ROUND(AVG(CAST([Mean TemperatureC] AS FLOAT)), 1) as avg_mean_temp_c FROM week06.london_2014
    UNION ALL SELECT 'Beijing', ROUND(AVG(CAST([Mean TemperatureC] AS FLOAT)), 1) FROM week06.beijing_pek_2014
    UNION ALL SELECT 'Brasilia', ROUND(AVG(CAST([Mean TemperatureC] AS FLOAT)), 1) FROM week06.brasilia_bsb_2014
    UNION ALL SELECT 'Cape Town', ROUND(AVG(CAST([Mean TemperatureC] AS FLOAT)), 1) FROM week06.capetown_cpt_2014
    UNION ALL SELECT 'Delhi', ROUND(AVG(CAST([Mean TemperatureC] AS FLOAT)), 1) FROM week06.delhi_del_2014
    UNION ALL SELECT 'Moscow', ROUND(AVG(CAST([Mean TemperatureC] AS FLOAT)), 1) FROM week06.moscow_svo_2014
    ORDER BY city

"""
answer("s07", SQL_07)
````

### SQL Task 8 · CTE + window ranking (top-1 per group)
For **London**, find the **hottest day of each month** (highest `Max TemperatureC`). If two days tie,
take the **earlier** date.

**Return:** `month_no`, `day` (a real **date**, convert `GMT` with `TRY_CONVERT(date, GMT)`), `max_temp`,
sorted by `month_no`. *Revisit: §10, §11.3.*

````python
SQL_08 = """
    WITH ranked AS (
            SELECT
                MONTH(TRY_CONVERT(date, GMT)) AS month_no,
                TRY_CONVERT(date, GMT) AS day,
                [Max TemperatureC] AS max_temp,
                ROW_NUMBER() OVER (PARTITION BY MONTH(TRY_CONVERT(date, GMT))
                        ORDER BY [Max TemperatureC] DESC,
                        TRY_CONVERT(date, GMT) ASC
                )  as rn

        FROM week06.london_2014
    )
    SELECT month_no,
            day,
            max_temp
    FROM ranked
    WHERE rn = 1
    ORDER BY month_no;
"""
answer("s08", SQL_08)
````

### SQL Task 9 · Running total
From `week09.comtrade_milk_uk_monthly_14`, compute the UK's **total milk exports per month** (exclude `'World'`)
and a **year-to-date running total**.

**Return:** `Period`, `exports_usd`, `ytd_exports_usd`, sorted by `Period`. *Revisit: §11.4–11.5.*

````python
SQL_09 = """
WITH monthly AS (
        SELECT
        Period,
        SUM([Trade Value (US$)]) AS exports_usd
    FROM week09.comtrade_milk_uk_monthly_14
    WHERE [Trade Flow] = 'Exports' AND Partner <> 'World'
    GROUP BY Period
)
    SELECT Period,
        exports_usd,
        SUM(exports_usd) OVER (ORDER BY Period ROWS UNBOUNDED PRECEDING) AS ytd_exports_usd
    FROM monthly
    ORDER BY Period
"""
answer("s09", SQL_09)
````
