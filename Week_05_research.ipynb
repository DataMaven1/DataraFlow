##**Evaluating LLM Logical Robustness Under Negation, Anchoring, and Irrelevant Context**

Author: Abdul-qayyum Omobolaji Yussuf | ID: DF2026-190 | Date: October 10, 2026



---


#📝 **Project Overview**

Large language models (LLMs) are increasingly used for tasks that require reasoning, yet a correct answer to a clean prompt does not show that a model will answer reliably when the wording changes or unrelated text is added. Existing studies have examined negation, irrelevant context, and anchoring-style suggestions separately, mostly on mathematics, code, or numerical tasks, with different models and scoring. This study compares all three on the same deductive reasoning problems. One hundred problems are generated with the JustLogic benchmark (answers are True, False, or Uncertain), and each is presented in eight controlled versions: the original; single, double, and equivalent negative wording; a wrong and a right suggested answer; and a short and a longer irrelevant passage. Three LLMs accessed through OpenRouter answer every version under a fixed prompt, giving 2,400 planned responses. Correct answers are worked out before any model is run and checked partly by an independent reviewer. Paired comparisons with bootstrap confidence intervals are used to measure the change in accuracy, answer flips, consistency under meaning-preserving changes, and whether models change their answer when single negation reverses the meaning.
