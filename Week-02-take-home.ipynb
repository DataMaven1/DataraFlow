- Create a base class `BankAccount` with `owner`, `balance`, and methods `deposit()` and `withdraw()`.
- Create a subclass `SavingsAccount` that adds an `interest_rate` and a method `add_interest()` which increases the balance.
- Demonstrate deposits, withdrawals, and interest application.

````python
# base class (parent class)
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.balance = balance

    def deposit(self, amount):
        if amount > 0:
            self.balance += amount
            print(f"{self.owner} deposited: ${amount}. New balance: ${self.balance}")
        else:
            print("Deposit amount must be positive.")

    def withdraw(self, amount):
        if amount > 0 and amount <= self.balance:
            self.balance -= amount
            print(f"{self.owner} withdrew: ${amount}. New balance: ${self.balance}")
        else:
            print("Withdrawal amount must be positive and less than or equal to the balance.")
# child class
class SavingsAccount(BankAccount):
    def __init__(self, owner, balance=0, interest_rate=0.03):
        #call the parent constructor to initialize owner and balance
        super().__init__(owner, balance)
        self.interest_rate = interest_rate

    def add_interest(self):
        # Calculate interest based on current balance
        interest = self.balance * self.interest_rate
        # add the interest to the balance
        self.balance += interest
        print(f"{self.owner} earned interest: ${interest}. New balance: ${self.balance}")

# instantiation
savings = SavingsAccount('Yussuf', 2000, 0.1)

savings.deposit(700)
savings.withdraw(450)
savings.add_interest()
````

- Create a class `CountDown` that takes a number `n` and counts down to 1.  
- Implement the iterator protocol (`__iter__()` and `__next__()` methods).  
- Use it in a `for` loop to print numbers in reverse order.

````python
class CountDown:
    ''' A simple countdown iterator class. '''
    def __init__(self, n):
        self.n = n

# Implement the __iter__() method to return the iterator object itself.
    def __iter__(self):
        return self

# Implement the __next__() method to return the next value in the countdown.
    def __next__(self):
        if self.n <= 0:
            raise StopIteration
        else:
            current = self.n
            self.n -= 1
            return current

# iterate through the countdown using a for loop 
count = CountDown(5)
for number in count:
    print(number)
````

- Create a custom iterator `FileLineReader` that takes a filename and reads one line at a time.  
- Implement `__iter__()` and `__next__()` so it can be used in a `for` loop.  
- Test it by reading from a sample `.txt` file and printing each line.

````python
class FileLineReader:
    def __init__(self, filename):
        self.file = open(filename, 'r')

    def __iter__(self):
        return self

    def __next__(self):
        line = self.file.readline()
        #if line is empty
        if not line:
            self.file.close()
            raise StopIteration
        return line.strip()

reader = FileLineReader('sample.txt')
for line in reader:
    print(line)
````

- Create an abstract class `Shape` with an abstract method `area()`.  
- Create subclasses `Rectangle` and `Circle` that implement the `area()` method.  
- Create a list of different shapes and loop through them, printing their area without checking their type.


````python
from abc import ABC, abstractmethod
import math

# Abstract base class 
class Shape(ABC):
    @abstractmethod
    def area(self):
        pass

# Rectangle class (child class) inherits from Shape
class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height

# Circle class (child class) inherits from Shape
class Circle(Shape):
    def __init__(self, radius):
        self.radius = radius

    def area(self):
        return math.pi * (self.radius ** 2)

# create instances of Rectangle and Circle
areas = [Rectangle(3, 8), Circle(5)]

# calculate and print areas using polymorphism
for shape in areas:
    print(f"Area of {type(shape).__name__}: {shape.area()}")
````

Create a class `Fibonacci` that generates Fibonacci numbers up to a given `n` using an iterator protocol.

````python
class Fibonacci:
    ''' a class fibonacci that generates nums up to a given n using an iter()'''
    def __init__(self, n):
        self.n = n
        self.a = 0
        self.b = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.a > self.n:
            raise StopIteration
        num = self.a
        self.a, self.b = self.b, self.b + self.a
        return num
# obj usage- printing fibonacci numbers up to 70
fibo = Fibonacci(70)

for num in fibo:
    print(num)
````

### 5. Extra Features

- Count the total number of books borrowed by all members (use a class-level variable).  

````python
class Book:
    def __init__(self, title, author, year):
        self.title = title
        self.author = author
        self.year = year
        self.available = True #tracks whether book is available

    def get_description(self):
        return f"The Book titled {self.title} is written by {self.author} in {self.year}"
        
#Ebook subclass from book
class Ebook(Book):
    def __init__(self, title, author, year, file_size, format):
        super().__init__(title, author, year)
        self.file_size = file_size
        self.format = format

    def get_description(self):
        return super().get_description() + f" [ebook; {self.file_size}MB, {self.format}]"

# child class from  Book class
class PrintedBook(Book):
    def __init__(self, title, author, year, weight, dimensions):
        super().__init__(title, author, year)
        self.weight = weight
        self.dimensions = dimensions

    def get_description(self):
        return super().get_description() + f" [ Printed: {self.weight}kg and {self.dimensions}]"

#Now the Member class and its subclass
class Member:
    total_borrowed = 0 # class-level variable shared across all

    def __init__(self, name, member_id):
        self.name = name
        self.member_id = member_id
        self._borrowed_books = [] # Protected attribute

    def borrow_book(self, book):
        if book.available:
            self._borrowed_books.append(book)
            book.available = False
            Member.total_borrowed += 1
            return f"{self.name} borrows {book.title}"
        return f"{book.title} is not available"

    def __iter__(self):
        return iter(self._borrowed_books)
    #setter
    
# first child class - StudentMember
class StudentMember(Member):
    def __init__(self, name, member_id, grade_level):
        super().__init__(name, member_id)
        self.grade_level = grade_level

    def borrow_book(self, book):
        if len(self._borrowed_books) < 3:
            return super().borrow_book(book)
        return f"Student has reached Borrowing limit"

# second class from Member
class TeacherMember(Member):
    def __init__(self, name, member_id, department):
        super().__init__(name, member_id)
        self.department = department

    def borrow_book(self, book):
        if len(self._borrowed_books) < 5:
            return super().borrow_book(book)
        return f"Teacher has reached borrowed limits"
    
#Library class
class Library:
    def __init__(self, library_name):
        # private attribute
        self.__library_name = library_name
        self.books = []

    @property
    def library_name(self) -> str:
        return self.__library_name

    @library_name.setter
    def library_name(self, name: str):
        if name.strip():
            self.__library_name = name
        else:
            raise ValueError("Library name cannot be empty.")
    
    def add_book(self, book):
        self.books.append(book)

    def search_books(self, keyword):
        keyword = keyword.lower()

        return [
            x for x in self.books
            if keyword in x.title.lower() or keyword in x.author.lower()
        ]    # return books whose title OR author contains keyword

    def __iter__(self):
        return iter(self.books)
````

````python
# Create library
library = Library("Central Library")

# Create books
book1 = Book("Dawn of Ages", "Abdul Yussuf", 2026)
book2 = Ebook("Data Science 101", "Winner", 2027, 4, "PDF")
book3 = PrintedBook("Mathematics Engineering", "Praise", 2023, 1.2, "24 x 16 cm")

# Create studnt and teacher members
student = StudentMember("Yussuf", "S099", "Senior")
teacher = TeacherMember("Mr. Qayyum", "T010", "Computer Science")

# Add books
library.add_book(book1)
library.add_book(book2)
library.add_book(book3)

# Test library name
print(library.library_name)

# Test library iterator
print("\nAll books:")
for book in library:
    print(book.get_description())

# Test borrowing
print("\nBorrowing:")
print(student.borrow_book(book1))
print(teacher.borrow_book(book2))

# Test member iterator
print("\nStudent's borrowed books:")
for book in student:
    print(book.title)

# Test searching
print("\nSearch results:")
for book in library.search_books("dawn"):
    print(book.title)

# Test total borrowed
print("\nTotal borrowed:", Member.total_borrowed)
````
